```python
from pathlib import Path
import pandas as pd

print("Pandas version:", pd.__version__)
print("Current working directory:", Path.cwd())

# Try to locate the project root automatically
cwd = Path.cwd()

possible_paths = [
    cwd / "ai" / "thermal_model" / "data" / "raw" / "Data.csv",
    cwd.parent / "data" / "raw" / "Data.csv",
    cwd / ".." / "data" / "raw" / "Data.csv",
]

DATA_PATH = None

for path in possible_paths:
    path = path.resolve()
    if path.exists():
        DATA_PATH = path
        break

print("Dataset path:", DATA_PATH)
print("Dataset found:", DATA_PATH is not None)
```

```python
encodings = [
    "utf-8-sig",
    "cp1250",
    "iso-8859-2",
    "cp1252",
    "latin1"
]

for enc in encodings:
    try:
        test_df = pd.read_csv(DATA_PATH, encoding=enc)
        print(
            f"SUCCESS: {enc} -> "
            f"{test_df.shape[0]} rows, {test_df.shape[1]} columns"
        )
    except Exception as e:
        print(f"FAILED: {enc} -> {type(e).__name__}")
```

```python
df = pd.read_csv(
    DATA_PATH,
    encoding="cp1250",
    low_memory=False
)

print("Dataset shape:", df.shape)
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

print("\nFirst 10 column names:")
print(df.columns[:10].tolist())

print("\nLast 10 column names:")
print(df.columns[-10:].tolist())

df.head()
```

```python
important_cols = ["e7", "e9t", "e9w", "e3"]

print("Column data types:")
print(df[important_cols].dtypes)

print("\nFirst 15 rows:")
display(df[important_cols].head(15))

for col in important_cols:
    print(f"\n--- {col} ---")
    print("Missing values:", df[col].isna().sum())
    print("Number of unique values:", df[col].nunique(dropna=False))
    print("Sample unique values:")
    print(df[col].drop_duplicates().head(20).tolist())
```

```python
# Check location and thermal-sensation distributions
print("e7 - Interview location:")
print(df["e7"].value_counts().sort_index())

print("\ne3 - Thermal sensation:")
print(df["e3"].value_counts().sort_index())


# Temporarily convert temperature and humidity to numeric
# Replace European decimal comma with decimal point
temp_numeric = pd.to_numeric(
    df["e9t"].astype(str).str.replace(",", ".", regex=False),
    errors="coerce"
)

humidity_numeric = pd.to_numeric(
    df["e9w"].astype(str).str.replace(",", ".", regex=False),
    errors="coerce"
)


print("\nTemperature (e9t):")
print("Conversion failures:", temp_numeric.isna().sum())
print("Minimum:", temp_numeric.min())
print("Maximum:", temp_numeric.max())

print("\nHighest 15 temperature values:")
print(temp_numeric.value_counts().sort_index(ascending=False).head(15))


print("\nHumidity (e9w):")
print("Conversion failures:", humidity_numeric.isna().sum())
print("Minimum:", humidity_numeric.min())
print("Maximum:", humidity_numeric.max())

print("\nHighest 15 humidity values:")
print(humidity_numeric.value_counts().sort_index(ascending=False).head(15))
```

```python
# Create a clean modelling subset without changing the original df
model_df = df[df["e7"].isin([2, 3])].copy()

# Keep only valid thermal sensation responses
model_df = model_df[model_df["e3"].between(1, 9)].copy()

# Convert European decimal commas to numeric values
model_df["temperature_c"] = pd.to_numeric(
    model_df["e9t"].astype(str).str.replace(",", ".", regex=False),
    errors="coerce"
)

model_df["relative_humidity"] = pd.to_numeric(
    model_df["e9w"].astype(str).str.replace(",", ".", regex=False),
    errors="coerce"
)

# Remove rows where temperature or humidity could not be converted
model_df = model_df.dropna(
    subset=["temperature_c", "relative_humidity"]
).copy()


print("Usable modelling rows:", len(model_df))

print("\nInterview location distribution:")
print(model_df["e7"].value_counts().sort_index())

print("\nThermal sensation distribution:")
print(model_df["e3"].value_counts().sort_index())

print("\nTemperature range:")
print(
    model_df["temperature_c"].min(),
    "to",
    model_df["temperature_c"].max()
)

print("\nHumidity range:")
print(
    model_df["relative_humidity"].min(),
    "to",
    model_df["relative_humidity"].max()
)

display(
    model_df[
        ["nr", "e7", "temperature_c", "relative_humidity", "e3"]
    ].head(10)
)
```

```python
def map_thermal_class(e3_value):
    if 1 <= e3_value <= 5:
        return "Not warm / OK"
    elif 6 <= e3_value <= 7:
        return "Warm"
    elif 8 <= e3_value <= 9:
        return "Too hot"
    else:
        return None


model_df["thermal_class"] = model_df["e3"].apply(map_thermal_class)


print("Thermal class distribution:")
print(model_df["thermal_class"].value_counts())


print("\nThermal class percentages:")
print(
    (
        model_df["thermal_class"]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
    )
)


print("\nCheck mapping from original e3 to new class:")
print(
    model_df.groupby(["e3", "thermal_class"])
    .size()
    .reset_index(name="count")
)


display(
    model_df[
        [
            "nr",
            "temperature_c",
            "relative_humidity",
            "e3",
            "thermal_class"
        ]
    ].head(15)
)
```

```python
print("Temperature summary:")
print(model_df["temperature_c"].describe())

print("\nHumidity summary:")
print(model_df["relative_humidity"].describe())


print("\nLowest 10 temperature records:")
display(
    model_df[
        ["nr", "e7", "temperature_c", "relative_humidity", "e3", "thermal_class"]
    ]
    .sort_values("temperature_c")
    .head(10)
)


print("\nHighest 10 temperature records:")
display(
    model_df[
        ["nr", "e7", "temperature_c", "relative_humidity", "e3", "thermal_class"]
    ]
    .sort_values("temperature_c", ascending=False)
    .head(10)
)


print("\nLowest 10 humidity records:")
display(
    model_df[
        ["nr", "e7", "temperature_c", "relative_humidity", "e3", "thermal_class"]
    ]
    .sort_values("relative_humidity")
    .head(10)
)


print("\nHighest 10 humidity records:")
display(
    model_df[
        ["nr", "e7", "temperature_c", "relative_humidity", "e3", "thermal_class"]
    ]
    .sort_values("relative_humidity", ascending=False)
    .head(10)
)
```

```python
# Check respondent IDs
print("Total rows:", len(model_df))
print("Unique respondent IDs:", model_df["nr"].nunique())

duplicate_ids = model_df["nr"].duplicated().sum()
print("Duplicate respondent IDs:", duplicate_ids)


# Check exact duplicates across the important modelling fields
important_model_cols = [
    "nr",
    "e7",
    "temperature_c",
    "relative_humidity",
    "e3",
    "thermal_class"
]

exact_duplicates = model_df.duplicated(
    subset=important_model_cols,
    keep=False
)

print("\nRows involved in exact duplicates:")
print(exact_duplicates.sum())


# Check repeated environmental conditions
condition_counts = (
    model_df
    .groupby(["temperature_c", "relative_humidity"])
    .size()
    .reset_index(name="count")
    .sort_values("count", ascending=False)
)

print("\nNumber of unique temperature-humidity combinations:")
print(len(condition_counts))

print("\nNumber of combinations appearing more than once:")
print((condition_counts["count"] > 1).sum())

print("\nTop 15 repeated environmental conditions:")
display(condition_counts.head(15))


# Show rows from the most repeated condition
top_condition = condition_counts.iloc[0]

same_condition_rows = model_df[
    (model_df["temperature_c"] == top_condition["temperature_c"]) &
    (model_df["relative_humidity"] == top_condition["relative_humidity"])
]

print(
    f"\nRows with most repeated condition "
    f"({top_condition['temperature_c']}°C, "
    f"{top_condition['relative_humidity']}% humidity):"
)

display(
    same_condition_rows[
        [
            "nr",
            "lokalizacja",
            "e7",
            "temperature_c",
            "relative_humidity",
            "e3",
            "thermal_class"
        ]
    ]
)
```

```python
print("City distribution:")
print(model_df["lokalizacja"].value_counts().sort_index())


print("\nCity distribution by thermal class:")
city_class_counts = pd.crosstab(
    model_df["lokalizacja"],
    model_df["thermal_class"]
)

print(city_class_counts)


print("\nCity distribution by thermal class (% within each city):")
city_class_percent = pd.crosstab(
    model_df["lokalizacja"],
    model_df["thermal_class"],
    normalize="index"
).mul(100).round(2)

print(city_class_percent)


print("\nEnvironmental summary by city:")

city_summary = (
    model_df
    .groupby("lokalizacja")
    .agg(
        rows=("nr", "count"),
        temp_min=("temperature_c", "min"),
        temp_mean=("temperature_c", "mean"),
        temp_max=("temperature_c", "max"),
        humidity_min=("relative_humidity", "min"),
        humidity_mean=("relative_humidity", "mean"),
        humidity_max=("relative_humidity", "max")
    )
    .round(2)
)

display(city_summary)


print("\nOriginal e3 distribution by city:")
print(
    pd.crosstab(
        model_df["lokalizacja"],
        model_df["e3"]
    )
)
```

```python
# Map the text target to an ordered numeric target
target_mapping = {
    "Not warm / OK": 0,
    "Warm": 1,
    "Too hot": 2
}

model_df["thermal_class_code"] = (
    model_df["thermal_class"]
    .map(target_mapping)
    .astype(int)
)


# Create a cleaner city label for later evaluation
city_mapping = {
    1: "Warsaw",
    2: "Madrid"
}

model_df["city"] = model_df["lokalizacja"].map(city_mapping)


# Keep only the columns needed for modelling and evaluation
prepared_df = model_df[
    [
        "nr",
        "city",
        "e7",
        "temperature_c",
        "relative_humidity",
        "e3",
        "thermal_class",
        "thermal_class_code"
    ]
].copy()


# Check the final dataset
print("Prepared dataset shape:", prepared_df.shape)

print("\nTarget code distribution:")
print(
    prepared_df["thermal_class_code"]
    .value_counts()
    .sort_index()
)

print("\nCheck target mapping:")
print(
    prepared_df[
        ["thermal_class", "thermal_class_code"]
    ]
    .drop_duplicates()
    .sort_values("thermal_class_code")
)

print("\nMissing values:")
print(prepared_df.isna().sum())


# Create processed directory if it does not exist
PROCESSED_DIR = DATA_PATH.parent.parent / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_PATH = PROCESSED_DIR / "thermosurvey_prepared.csv"

# Save clean modelling dataset
prepared_df.to_csv(
    OUTPUT_PATH,
    index=False,
    encoding="utf-8"
)

print("\nSaved to:")
print(OUTPUT_PATH)

print("\nFile exists:", OUTPUT_PATH.exists())


display(prepared_df.head(10))
```

```python
# Reload the processed dataset from disk
check_df = pd.read_csv(OUTPUT_PATH)

print("Reloaded shape:", check_df.shape)

print("\nColumn names:")
print(check_df.columns.tolist())

print("\nData types:")
print(check_df.dtypes)

print("\nMissing values:")
print(check_df.isna().sum())

print("\nTarget distribution:")
print(
    check_df["thermal_class_code"]
    .value_counts()
    .sort_index()
)

print("\nTemperature range:")
print(
    check_df["temperature_c"].min(),
    "to",
    check_df["temperature_c"].max()
)

print("\nHumidity range:")
print(
    check_df["relative_humidity"].min(),
    "to",
    check_df["relative_humidity"].max()
)

display(check_df.head(10))
```
